# Prompt Datasets

This notebook loads and showcases the three main prompt datasets used in the Zigzag Diffusion Sampling paper: Pick-a-Pic, DrawBench, and GenEval.

```python
from pathlib import Path

import pandas as pd

from zigzag_project.dataset_utils import load_drawbench_prompts, load_pickapic_prompts

PROJECT_ROOT = Path.cwd().parent
```

# Pick-a-Pic

We followed the Pick-a-Pic reference from the ZigZag Diffusion Sampling paper:

*Yuval Kirstain, Adam Polyak, Uriel Singer, Shahbuland Matiana, Joe Penna, and Omer Levy. Picka-pic: An open dataset of user preferences for text-to-image generation. Advances in Neural Information Processing Systems, 36:36652–36663, 2023.*

However, we discovered that the original `pickapick_v1` and `pickapick_v2` datasets that were said to be available via HuggingFace are no longer accesible. Instead, we downloaded a set of 200 selected Pick-a-Pic prompts from the author, available at https://huggingface.co/datasets/yuvalkirstain/PickaPic-selected-prompts. Thus, we cannot guarantee these are the exact same prompts used in the ZigZag paper, but they are from the same dataset.

```python
PICKAPIC_CSV = PROJECT_ROOT / "data/prompts/pick-a-pic.csv"
print(f"PICKAPIC_CSV.exists() = {PICKAPIC_CSV.exists()}\n")

# The paper only uses the first 100 Pick-a-Pic prompts.
pickapic = load_pickapic_prompts(csv_path=PICKAPIC_CSV, limit=100)
pickapic.info(verbose=False)
pickapic.head(10)
```

# DrawBench

We followed the DrawBench reference from the ZigZag Diffusion Sampling paper:

*Chitwan Saharia, William Chan, Saurabh Saxena, Lala Li, Jay Whang, Emily L Denton, Kamyar Ghasemipour, Raphael Gontijo Lopes, Burcu Karagol Ayan, Tim Salimans, et al. Photorealistic text-to-image diffusion models with deep language understanding. Advances in neural information processing systems, 35:36479–36494, 2022.*

This lead us to a Google sheet file containing the 200 Drawbench prompts, available at https://docs.google.com/spreadsheets/d/1y7nAbmR4FREi6npB1u-Bo3GFdwdOPYJc617rBOxIRHY/edit#gid=0.


```python
DRAWBENCH_CSV = PROJECT_ROOT / "data/prompts/draw-bench.csv"
print(f"DRAWBENCH_CSV = {DRAWBENCH_CSV}")
print(f"DRAWBENCH_CSV.exists() = {DRAWBENCH_CSV.exists()}\n")

drawbench = load_drawbench_prompts(csv_path=DRAWBENCH_CSV, limit=None)
drawbench.info(verbose=False)
drawbench.head(10)
```
